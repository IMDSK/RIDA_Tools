```python
import os
import ipywidgets as widgets
from ipywidgets import HBox, Label, Button
from IPython.display import display
from ipyfilechooser import FileChooser

import ShapeProcess
import FileChecker
import CropProcess
import FolderMover
import subprocess

# Global variables
Drive = "/Users/imdsk/RIDA_Tools/Sentinel_Process/"
Image = "/Users/imdsk/RIDA_Tools/Sentinel_Process/Image"
Image_Pre = "/Users/imdsk/RIDA_Tools/Sentinel_Process/Image_Pre"
Image_Finish = os.path.join(Drive, "Image_Finish")
Image_Missing = os.path.join(Drive, "Image_Missing")
Output = os.path.join(Drive, "Output")
Rtbcon = os.path.join(Drive, "Raster_BurnCon")
Rtbreg = os.path.join(Drive, "Raster_BurnReg")
RtbShape = os.path.join(Drive, "Raster_BurnShape")
RtbLevel = os.path.join(Drive, "Raster_BurnLevel")

def process_data(selected_folder):
    csv_file = os.path.join(selected_folder, "Coordinate.csv")
    ShapeProcess.execute_script(csv_file, selected_folder)
    response = FileChecker.process_files(selected_folder)
    print(f"Data processed for folder: {selected_folder}")
    print(response)
    
    shp_path = os.path.join(selected_folder, "SHP")
    
    # Create a list of directories to create folders in
    target_dirs = [Image_Finish, Image_Missing, Output, Rtbcon, Rtbreg, RtbShape, RtbLevel]
    
    for area_dir in os.listdir(shp_path):
        area_path = os.path.join(shp_path, area_dir)
        if os.path.isdir(area_path):
            CropProcess.crop_and_show_images(selected_folder, area_path, 'Before', 'Before_Cropped')
            CropProcess.crop_and_show_images(selected_folder, area_path, 'After', 'After_Cropped')
            
            # Create folders with the same name as the cropped image folder in the target directories
            for target_dir in target_dirs:
                folder_name = os.path.basename(area_path)
                new_folder_path = os.path.join(target_dir, folder_name)
                os.makedirs(new_folder_path, exist_ok=True)
            
    FolderMover.move_folders(selected_folder) 

    script_path = "/Users/imdsk/RIDA_Tools/SentinelProcess.py"
    subprocess.run(["python", script_path])

def submit_folder(b):
    folder_path = fc.selected_path
    if folder_path and os.path.isdir(folder_path):
        folder_label.value = f"Selected folder: {folder_path}"
        process_data(folder_path)
    else:
        folder_label.value = "Invalid or no folder selected"

# Create a FileChooser widget for directory selection
fc = FileChooser('/')
fc.show_only_dirs = True
display(fc)

submit_button = Button(description="Process Folder")
submit_button.on_click(submit_folder)

folder_label = Label(value="No folder selected")
display(HBox([submit_button, folder_label]))
```
