```python
import folium

points = [
    [40.7128, -74.0059],  # New York City
    [51.5074, -0.1278],   # London
    [48.8566, 2.3522],    # Paris
    [35.6895, 139.6917],  # Tokyo
    [40.7128, -74.0059]   # Back to New York City (to close the polygon)
]

# Create a map object
map = folium.Map()

# Create a polygon from the points
polygon = folium.vector_layers.Polygon(
    locations=points,
    color='red',
    fill=True,
    fill_color='red',
    fill_opacity=0.4
)

# Add the polygon to the map
polygon.add_to(map)

# Display the map
map
```
